1.- Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.

```python
def cadenaTexto (palabra):
    miDiccionario = {}

    # Recorremos la "palabra" letra por letra
    for letra in palabra: 
        # si una de esas letras esta en blanco que no haga nada 
        if letra == " ":
            continue
        # si la letra esta en el diccionario que le sume un punto
        if letra in miDiccionario:
            miDiccionario[letra] += 1
        # si no esta que cree una key y le asignamos el valor de 1
        else:
            miDiccionario[letra] = 1

    return miDiccionario
        

usuario_input = input("Ingrese una palabra")
resultado = cadenaTexto(usuario_input)


print(resultado)
```

2.- Dada una lista de números, obtén una nueva lista con el doble de cada valor. Usa la función map().

```python
listaPrueba = [1,2,3,4,5]

def duplicarNum(num):
    # devolvemos el numero * 2
    return num*2

# usamos map para aplicar una funcion a cada elemento, despues le ponemor un list() ya que nos devuelve un objeto map y nosotros queremos una lista
print(list(map(duplicarNum,listaPrueba)))
```

3.- Escribe una función que tome una lista de palabras y una palabra objetivo como parámetros. La función debe devolver una lista con todas las palabras de la lista original que contengan la palabra objetivo.

```python
def filtrar_palabras(lista_original, palabra_objetivo):
    coincidencias = []
    
    # Recorremos la lista original palabra por palabra
    for palabra in lista_original:
        # Si la palabra objetivo está dentro de la palabra actual la guardamos en "coincidencias[]"
        if palabra_objetivo in palabra:
            coincidencias.append(palabra)
            
    return coincidencias

mis_palabras = ["girasol", "casa", "solera", "perro", "sol", "marisol"]
objetivo = "sol"

resultado = filtrar_palabras(mis_palabras, objetivo)
print(resultado) 
```

4.- Genera una función que calcule la diferencia entre los valores de dos listas. Usa la función map().

```python
def restar_valores(a, b):
    return a - b

lista_1 = [10, 20, 30, 40]
lista_2 = [3,  5,  10, 12]

resultado = list(map(restar_valores, lista_1, lista_2))

print(resultado)
```

5.- Escribe una función que tome una lista de números como parámetro y un valor opcional nota_aprobado (por defecto 5). La función debe calcular la media de los números en la lista y determinar si la media es mayor o igual que nota_aprobado. Si es así, el estado será "aprobado"; de lo contrario, "suspenso". La función debe devolver una tupla que contenga la media y el estado.

```python
def mediaAlumnos(listaAlumnos, nota_aprobado=5):
    sumaPuntos = 0
    
    # recorremos la lista de notas y las sumamos en "sumaPuntos"
    for notas in listaAlumnos:
        sumaPuntos += notas
        
    media = sumaPuntos / len(listaAlumnos)
    
    # una vez pasado la media la comprobamos con la nota de aprobado
    if media >= nota_aprobado:
        estado = "aprobado"
    else:
        estado = "suspenso"
        
    # devolvemos la media y el estado
    return (media, estado)

listaNotas = [7,7,8,6,4,8]
notaAprobado = 5

print(mediaAlumnos(listaNotas,notaAprobado))
```

6.- Escribe una función que calcule el factorial de un número de manera recursiva.

```python
def factorial(num):
    contador = 1
    resultado = 1

# usamos un bucle while para ir de numero en numero
    while contador <= num:
        resultado *= contador
        contador += 1     
    
    return resultado

resultadofact = factorial(10)
print(f"El resultado es {resultadofact}")
```

7.- Genera una función que convierta una lista de tuplas a una lista de strings. Usa la función map().

```python
def convertidorTuplas(tupla):
    # usamos el join para unir cada string con un espacio entre cada una 
    return " ".join(tupla)

lista_de_tuplas = [("Hola", "Mundo"), ("Me", "gusta", "Python"), ("Ejercicio", "siete")]

lista_de_string = list(map(convertidorTuplas,lista_de_tuplas))

print(lista_de_string)
```

8.- Escribe un programa que pida al usuario dos números e intente dividirlos. Si el usuario ingresa un valor no numérico o intenta dividir por cero, maneja esas excepciones de manera adecuada y muestra un mensaje indicando si la división fue exitosa o no. 

```python
def division():
    try:
        # 1. Pedimos los números al usuario y los transformamos a float (por si usan decimales)
        num1 = float(input("Introduce el primer número: "))
        num2 = float(input("Introduce el segundo número: "))
        
        # 2. Intentamos hacer la división
        resultado = num1 / num2
        
    except ValueError:
        # Este bloque se ejecuta si el usuario introduce letras en vez de números
        print("Error: Has ingresado un valor no numérico.")
        print("La división NO fue exitosa.")
        
    except ZeroDivisionError:
        # Este bloque se ejecuta si el segundo número es 0
        print("Error: No se puede dividir entre cero.")
        print("La división NO fue exitosa.")
        
    else:
        # Este bloque solo se ejecuta si el bloque 'try' funcionó a la perfección
        print("¡La división fue exitosa!")
        print(f"El resultado de {num1} / {num2} es: {resultado}")

# Llamamos a la función para probarla

division()
```
