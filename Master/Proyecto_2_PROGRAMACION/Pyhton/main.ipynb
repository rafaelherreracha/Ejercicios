1.- Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.

```python
def cadenaTexto (palabra):
    miDiccionario = {}

    # Recorremos la "palabra" letra por letra
    for letra in palabra: 
        # si una de esas letras esta en blanco que no haga nada 
        if letra == " ":
            continue
        # si la letra esta en el diccionario que le sume un punto
        if letra in miDiccionario:
            miDiccionario[letra] += 1
        # si no esta que cree una key y le asignamos el valor de 1
        else:
            miDiccionario[letra] = 1

    return miDiccionario
        

usuario_input = input("Ingrese una palabra")
resultado = cadenaTexto(usuario_input)


print(resultado)
```

5.- Escribe una función que tome una lista de números como parámetro y un valor opcional nota_aprobado (por defecto 5). La función debe devolver una tupla que contenga la media y el estado.

```python
def mediaAlumnos(listaAlumnos, nota_aprobado=5):
    sumaPuntos = 0
    
    # recorremos la lista de notas y las sumamos en "sumaPuntos"
    for notas in listaAlumnos:
        sumaPuntos += notas
        
    media = sumaPuntos / len(listaAlumnos)
    
    # una vez pasado la media la comprobamos con la nota de aprobado
    if media >= nota_aprobado:
        estado = "aprobado"
    else:
        estado = "suspenso"
        
    # devolvemos la media y el estado
    return (media, estado)

listaNotas = [7,7,8,6,4,8]
notaAprobado = 5

print(mediaAlumnos(listaNotas,notaAprobado))
```

6.- Escribe una función que calcule el factorial de un número de manera recursiva.

```python
def factorial(num):
    contador = 1
    resultado = 1

# usamos un bucle while para ir de numero en numero
    while contador <= num:
        resultado *= contador
        contador += 1     
    
    return resultado

resultadofact = factorial(10)
print(f"El resultado es {resultadofact}")
```

```python
## Definimos la excepción personalizada
class ListaVaciaError(Exception):
    """Excepción lanzada cuando la lista de números está vacía."""
    pass

## función para calcular el promedio
def calcular_promedio(numeros):
    # Si la lista está vacía, lanzamos nuestra excepción personalizada
    ## Aqui usamos raise para lanzar una excepcion a voluntad
    if not numeros:
        raise ListaVaciaError("No se puede calcular el promedio de una lista vacía.")
    
    return sum(numeros) / len(numeros)


listas_de_prueba = [
    [10, 20, 30, 40], 
    []                
]

for lista in listas_de_prueba:
    try:
        print(f"\nIntentando calcular el promedio de: {lista}")
        resultado = calcular_promedio(lista)
        print(f"¡Éxito! El promedio es: {resultado}")
        
    except ListaVaciaError as error:
        print(f"Error controlado: {error}")
```

28.-Crea una función que busque y devuelva el primer elemento duplicado en una lista dada.

```python
def primer_duplicado(lista):
    vistos = set()
    for elemento in lista:
        if elemento in vistos:
            return elemento
        vistos.add(elemento)
    return None  # Si no hay duplicados

# Ejemplo de uso:
print(primer_duplicado([1, 2, 3, 2, 4, 1]))  # 2
```

```python
def enmascarar_variable(variable):
    cadena = str(variable)
    if len(cadena) <= 4:
        return cadena
    return "#" * (len(cadena) - 4) + cadena[-4:]

# Ejemplo de uso:
print(enmascarar_variable("123456789"))  # #####6789
```

30.-Crea una función que determine si dos palabras son anagramas, es decir, si están formadas por las mismas letras pero en diferente orden.

```python
def son_anagramas(palabra1, palabra2):
    # Quitamos espacios y pasamos a minúsculas para una comparación limpia
    p1 = palabra1.replace(" ", "").lower()
    p2 = palabra2.replace(" ", "").lower()
    return sorted(p1) == sorted(p2)

# Ejemplo de uso:
print(son_anagramas("Raza", "Zara"))  # True
```

34.-Crea la clase Arbol

```python
class Arbol:
    def __init__(self):
        self.tronco = 1
        self.ramas = []
        
    def crecer_tronco(self):
        self.tronco += 1
        
    def nueva_rama(self):
        self.ramas.append(1)
        
    def crecer_ramas(self):
        self.ramas = [rama + 1 for rama in self.ramas]
        
    def quitar_rama(self, posicion):
        if 0 <= posicion < len(self.ramas):
            self.ramas.pop(posicion)
            
    def info_arbol(self):
        return {
            "longitud_tronco": self.tronco,
            "numero_ramas": len(self.ramas),
            "longitudes_ramas": self.ramas
        }

# --- Caso de Uso ---
mi_arbol = Arbol()                  # a. Crear un árbol.
mi_arbol.crecer_tronco()            # b. Hacer crecer el tronco una unidad.
mi_arbol.nueva_rama()               # c. Añadir una nueva rama.
mi_arbol.crecer_ramas()             # d. Hacer crecer todas las ramas una unidad.
mi_arbol.nueva_rama()               # e. Añadir dos nuevas ramas...
mi_arbol.nueva_rama()               #    ... (segunda rama)
mi_arbol.quitar_rama(2)             # f. Retirar la rama situada en la posición 2.
print(mi_arbol.info_arbol())        # g. Obtener información sobre el árbol.
# Resultado esperado: {'longitud_tronco': 2, 'numero_ramas': 2, 'longitudes_ramas': [2, 1]}
```

35.-Crea la clase UsuarioBanco

```python
class UsuarioBanco:
    def __init__(self, nombre, saldo, cuenta_corriente=True):
        self.nombre = nombre
        self.saldo = saldo
        self.cuenta_corriente = cuenta_corriente
        
    def agregar_dinero(self, cantidad):
        self.saldo += cantidad
        
    def retirar_dinero(self, cantidad):
        if cantidad > self.saldo:
            raise ValueError(f"{self.nombre} no tiene saldo suficiente.")
        self.saldo -= cantidad
        
    def transferir_dinero(self, destino, cantidad):
        try:
            self.retirar_dinero(cantidad)
            destino.agregar_dinero(cantidad)
        except ValueError as e:
            print(f"Transferencia fallida: {e}")

# --- Caso de Uso ---
# a. Crear dos usuarios
alicia = UsuarioBanco("Alicia", 100, True)
bob = UsuarioBanco("Bob", 50, True)

bob.agregar_dinero(20)          # b. Agregar 20 unidades a Bob.
bob.transferir_dinero(alicia, 80) # c. Transferir 80 unidades de Bob a Alicia. (Dará error controlado porque Bob tiene 70)
alicia.retirar_dinero(50)       # d. Retirar 50 unidades de Alicia.

print(f"Saldo Alicia: {alicia.saldo}, Saldo Bob: {bob.saldo}")
```

36.-Crea una función llamada procesar_texto

```python
def contar_palabras(texto):
    diccionario = {}
    for palabra in texto.split():
        palabra = palabra.lower().strip(",.")
        diccionario[palabra] = diccionario.get(palabra, 0) + 1
    return diccionario

def reemplazar_palabras(texto, original, nueva):
    return texto.replace(original, nueva)

def eliminar_palabra(texto, palabra):
    palabras = texto.split()
    palabras_filtradas = [p for p in palabras if p.strip(",.") != palabra]
    return " ".join(palabras_filtradas)

def procesar_texto(texto, opcion, *args):
    if opcion == "contar":
        return contar_palabras(texto)
    elif opcion == "reemplazar":
        return reemplazar_palabras(texto, args[0], args[1])
    elif opcion == "eliminar":
        return eliminar_palabra(texto, args[0])
    else:
        return "Opción no válida"

# --- Caso de uso ---
texto_prueba = "hola mundo hola"
print(procesar_texto(texto_prueba, "contar"))  # {'hola': 2, 'mundo': 1}
print(procesar_texto(texto_prueba, "reemplazar", "hola", "adiós"))  # "adiós mundo adiós"
```

```python
def momento_del_dia(hora):
    if 6 <= hora < 13:
        return "Es de mañana/día"
    elif 13 <= hora < 21:
        return "Es de tarde"
    elif (21 <= hora <= 24) or (0 <= hora < 6):
        return "Es de noche"
    else:
        return "Hora no válida"

# Ejemplo de uso:
print(momento_del_dia(16))  # Es de tarde
```

39.-Escribe una función que tome dos parámetros: figura (una cadena que puede ser "rectangulo", "circulo" o "triangulo") y datos (una tupla con los datos necesarios para calcular el área de la figura).

```python
import math

def calcular_area(figura, datos):
    if figura == "rectangulo":
        base, altura = datos
        return base * altura
    elif figura == "circulo":
        radio = datos[0]
        return math.pi * (radio ** 2)
    elif figura == "triangulo":
        base, altura = datos
        return (base * altura) / 2
    else:
        return "Figura no soportada"

# Ejemplo de uso:
print(calcular_area("rectangulo", (5, 4)))  # 20
```

```python
def calcular_monto_compra():
    try:
        precio_original = float(input("Introduce el precio original del artículo: "))
        tiene_cupon = input("¿Tienes un cupón de descuento? (sí/no): ").strip().lower()
        
        if tiene_cupon == "sí" or tiene_cupon == "si":
            valor_cupon = float(input("Introduce el valor del cupón de descuento: "))
            
            if valor_cupon > 0:
                precio_final = precio_original - valor_cupon
                # Si el cupón es mayor que el precio, el precio final no debería ser negativo
                precio_final = max(0.0, precio_final) 
            else:
                print("Cupón no válido (debe ser mayor a cero). No se aplicará descuento.")
                precio_final = precio_original
        else:
            precio_final = precio_original
            
        print(f"El precio final de la compra es: {precio_final:.2f}")
        
    except ValueError:
        print("Error: Has introducido un formato de número incorrecto.")

# Ejemplo de uso:
# calcular_monto_compra()
```
